# 00 · Setup
Crea la estructura en Unity Catalog para el pipeline medallón:
schemas `landing`, `bronze`, `silver`, `gold` y el volume `landing.raw`.
Se ejecuta una sola vez.

```sql
%sql
SHOW CATALOGS;
```

```sql
%sql
-- Celda 1: catálogo de trabajo
USE CATALOG workspace;
```

```sql
-- Celda 2: capas de la arquitectura medallón
CREATE SCHEMA IF NOT EXISTS landing COMMENT 'Zona de aterrizaje: archivos crudos tal como llegan desde la fuente';
CREATE SCHEMA IF NOT EXISTS bronze  COMMENT 'Capa bronze: datos crudos ingeridos + metadata de ingesta';
CREATE SCHEMA IF NOT EXISTS silver  COMMENT 'Capa silver: datos limpios, tipados, deduplicados y validados';
CREATE SCHEMA IF NOT EXISTS gold    COMMENT 'Capa gold: modelos de negocio listos para consumo (BI, Genie)';
```

```sql
-- Celda 3: volume para los archivos de entrada
CREATE VOLUME IF NOT EXISTS landing.raw
COMMENT 'Archivos JSON que simulan la llegada incremental de datos desde sistemas origen';
```

```sql
-- Celda 4: verificación
SHOW SCHEMAS IN workspace;
```
